```python
import pandas as pd 
import numpy as numpy
from sklearn.metrics import cohen_kappa_score as ck
```

```python
ck_result = {}
for i in range(21):
    chapter = i+1
    df = pd.read_csv(f'./table/tab_chapter{chapter}.csv')
    ck_result[i+1] = ck(y1=df['ED'].values, y2=df['IPD'].values)

pd.DataFrame.from_dict(ck_result, orient='index', columns=['cohen_kappa_score'])
```
