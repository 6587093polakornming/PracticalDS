# Import Library

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
def transform_to_array(dataframe: pd.DataFrame):
    return dataframe.to_numpy().transpose()
```

```python
def evaluate_pvalue(pvalue: float, alpha=0.05) -> str:
    return "Reject H0" if pvalue < alpha else "NOT Reject H0"

# Reject H0     -> the difference in the distribution of the observed data is significant
# NOT Reject H0 -> the difference in the distribution of the observed data is NOT significant
```

```python
def generate_description(pvalue, alpha=0.05):
    output = "the difference in the distribution of the observed data is"
    output.join(" significant") if pvalue < alpha else output.join(" NOT significant")
    return output
```

# Create Result Report

```python
report_df = pd.DataFrame(columns=['Test Name','statistic','P-Value','Result(Alpha=0.05)', 'Description'])
report_df
```

```python
overall_table = pd.read_csv('table/overall_tab.csv')
overall_table = overall_table.drop(["chapter"], axis=1)
overall_table
```

```python
transformed_data = transform_to_array(overall_table)
print(transformed_data)
results = stats.chi2_contingency(transformed_data)

pvalue_result = results.pvalue
print(results)
print("pvalue:",pvalue_result)
print("statistic:",results.statistic)
print(evaluate_pvalue(pvalue_result))
```

```python
result_hypothesis = evaluate_pvalue(results.pvalue)
results_des = generate_description(results.pvalue)
```

```python
new_row = {
    'Test Name': 'overall_chapter',
    'statistic': results.statistic,
    'P-Value': results.pvalue,
    'Result(Alpha=0.05)': result_hypothesis, 
    'Description': results_des
}

report_df = pd.concat([report_df, pd.DataFrame([new_row])], ignore_index=True)
report_df
```

```python
for i in range(1,22):
    table = pd.read_csv(f'table/tab_chapter{i}.csv')
    table = table.drop(["icd_code"], axis=1)
    
    transformed_data = transform_to_array(table)
    results = stats.chi2_contingency(transformed_data)

    statistic = results.statistic
    pvalue = results.pvalue
    result_hypothesis = evaluate_pvalue(pvalue)
    results_des = generate_description(pvalue)

    new_row = {
        'Test Name': f'tab_chapter{i}',
        'statistic': statistic,
        'P-Value': pvalue,
        'Result(Alpha=0.05)': result_hypothesis, 
        'Description': results_des
    }
    report_df = pd.concat([report_df, pd.DataFrame([new_row])], ignore_index=True)
```
